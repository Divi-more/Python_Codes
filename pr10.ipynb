# *Exception:-

    ::Exception: Error in the code that occurs during execution (runtime) of program, causing normal flow of program to stop.

    *Without Exception Handling: - program terminates immediately
                                 - remaining code isn't executed
                                 - poor user experience 

    *With Exception Handling:    - exception does not happen

    # try-expect(catch) block or keywords:-
        * try     - code that may have error in it
        * expect  - handle the error
        * else    - show result if there's no error
        * finally - always execute

```python
#ZeroDivisionError

num1 = int(input("Enter num1: "))
num2 = int(input("Enter num2: "))

try: 
    res = num1 / num2
    
except ZeroDivisionError as err:
    print("Error: ",err)
    
else:
    print("Result: ", res)
    
finally:
    print("Division completed...")
```

```python
# ValueError: wrong value
# num1 = int(input("Enter num1: "))
# num2 = int(input("Enter num2: "))

# num1 = int("10")
# num2 = int("20")

try: 
    num1 = int(input("Enter num1: "))
    num2 = int(input("Enter num2: "))

    res = num1 + num2
    
except ValueError as err:
    print("Please enter valid number: \n ",err)
    
else:
    print("Result: ", res)
    
```

```python
# TypeError: when we give wrong data type

num1 = int(input("Enter num1: "))
num2 = input("Enter num2: ")
    
try: 
    # num1 = int(input("Enter num1: "))
    # num2 = input("Enter num2: ")
    res = num1 + num2
    
except TypeError as err:
    print("Wrong data type of given values: \n ",err)
    
else:
    print("Result: ", res)
```

```python
# IndexError
list1 = [10,20,30,40,50]

try:
    temp = list1[5]

except IndexError as err:
    print("Value not found: ", err)
    
else:
    print(temp)
```

```python
# KeyError

dict1 = {"name":"Divya", "marks":90}

try:
    user = dict1["roll"]
    
except KeyError as err:
    print("Key does not exist: ", err)
    
else:
    print("Result:",user)
```
