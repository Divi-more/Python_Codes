```python
"""
        *
       * *
      * * *
"""

col = 3
row = 1
space = " "
spacecount = 2

for i in range(col):        # outer loop run till 3
    for j in range(row):    # inner loop run till 1
        print(space * spacecount, "*", end="")
        spacecount -= 1
    print(" ")
    row += 1
```
