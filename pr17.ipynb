# getter - setter methods:
    1. Getter:-
        - method used to read/access value of an attribute.
    
    2. Setter:-
        - method used to update value of an attribute.

```python
class bank:
    
    def __init__(self):
        self.name = "DemoUser"
        self.acc_num = 777888
        self.__pin = 2012   # private
    
    # getter    
    def get_pin(self):
        print("getter method:-\nReturns the private value to this method &\nMethod can be accessed anywhere in code later\n\n")
        return self.__pin
    
    # setter
    def set_pin(self, pin):
        print("setter method:-\nChange values to private")
        self.__pin = pin
        
class sub_branch(bank):
    def display(self):
        print(f"User Name: {self.name} \nAcc Num: {self.acc_num}\nUser PIN: {self.get_pin()}")
        # print(f" User PIN: {self.__pin}")
        
    def update_pin(self,pin):
        self.set_pin(pin)
        
        
obj = sub_branch()
obj.set_pin(7576)
obj.display()
# obj.get_pin()
```

# Types of Inheritance:
    1. Single
    2. Multi-level
    3. Multiple
    4. Hierarchical
    5. Hybrid

```python
# Single Inheritance:-
class parent:
    
    def show(self):
        print("Parent class")
        
class child(parent):
    
    def display(self):
        print("Child class")

obj = child()
obj.show()
obj.display()
```

```python
# Multi-level Inheritance:-
class grandparent:
    
    def temp(self):
        print("Grand-Parent class")
        
class parent(grandparent):
    
    def temp(self):
        super().temp()
        print("Parent class")

class child(parent):
    
    def temp(self):
        super().temp()
        print("Child class")

obj = child()
obj.temp()
```

```python
# Multiple Inheritance:-

class parent1:
    def pOne(self):
        print("Parent 1")
        
class parent2:
    def pTwo(self):
        print("Parent 2")
        
class child(parent1, parent2):
    pass

obj = child()
obj.pOne()
obj.pTwo()
```

```python
# Hierarchical Inheritance:-

class parent:
    def temp(self):
        print("Parent class")
        
class child1(parent):
    def show(self):
        print("Child 1 class")

class child2(parent):
    pass

obj = child2()
obj.temp()
print()

obj1=child1()
obj1.temp()
obj1.show()
```

```python
# Hybrid Inheritance:-

class grandparent:
    def grad_m(self):
        print("Grand Parent class")
        
class parent(grandparent):  # single
    def parent_m(self):
        print("Parent class")
        
class child(parent):        # multi-level
    def child_m(self):
        print("Child class")
        
class demo:
    def demo_m(self):
        print("Demo class")
        
class temp(child, demo):    # multiple
    pass

obj = temp()
obj.grad_m()
obj.parent_m()
obj.child_m()
obj.demo_m()
```
