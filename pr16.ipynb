# Inheritance:-
    - allows one class to accquire th attributes and methods of another class.
    - parent and child class.
    - child class can reuse code of existing parent class.

### Types:
    1. single:(A->B)
    2. multi-level: (A->B->C)
    3. multiple: (A and B -> C)
    4. hierarchical
    5. hybrid

```python
# Inheritance:

class Parent:
    
    def parent_method(self):
        print("This is Parent class.")
        
    def greet(self):
        print("Hello")

class Child(Parent):                #inherit the Parent class
    
    def child_method(Self):
        print("This is Child class.")
 
obj1 = Child()

obj1.greet()
obj1.parent_method()
obj1.child_method()




```

```python
# Inheritance with constructor :
# If there's no child constructor, parent constructor gets called.

class Parent:
    
    def __init__(self):
        print("This is Parent class Constructor")
    
    def parent_method(self):
        print("This is Parent class.")

class Child(Parent):
        
    def __init__(self):
        print("This is Child class Constructor")
            
    def child_method(Self):
        print("This is Child class.")
 
obj1 = Child()
obj1.parent_method()
obj1.child_method()
```

super keyword:-

Commonly used to Parent class constructors and methods in any child class.

```python
class One:
    def greet(self):
        print("Hello from Parent class.")
        
class Two(One):
    def greet(self):
        super().greet()
        print("Hello from Child Class.")
        
obj1 = Two()
obj1.greet()
```

```python
class Parent:
    
    def __init__(self):
        print("This is Parent class Constructor")
    
    def parent_method(self):
        print("This is Parent class.")

class Child(Parent):
        
    def __init__(self):
        super().__init__()
        print("This is Child class Constructor")
            
    def child_method(Self):
        print("This is Child class.")
 
obj1 = Child()
# obj1.parent_method()
# obj1.child_method()
```

### Method Overriding: 
    - same method in different classes.


```python

class upi:
    def Pay(self):
        print("Payment using upi.")
        
class NetBanking:
    def Pay(self):
        print("Payment using net bankikg.")
        
class CreditCard:
    def Pay(self):
        print("Payment using credit cards.")
        
obj1 = upi()
obj1.Pay()

obj2 = NetBanking()
obj2.Pay()

obj3 = CreditCard()
obj3.Pay()
```

# Access Modifiers:
    - tell where variables can be accessible

    1. public - accessible anywhere (in other class if inherited parent), outside classes using 
                object
    2. protected - accessible in it's class only and subclass i.e., the one which has inherit 
                   it's parent class

    - demo   : public
    - _demo  : protected 
    - __demo : private
