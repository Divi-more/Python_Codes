### Function & Method Revision:
    
    # Function: when defined outside class, it's function.
    # Method: when defined inside class, it's method
    # def __init__: t's a special method/ cinstructor.

```python
class revision:
    def func(self,name,marks):
        self.name = name
        self.marks = marks
        
    def show(self):
        print(self.name,"\n", self.marks)
        
rev = revision()
rev.func("David", 95)
rev.show()
```

```python
class revision:
    def __init__(self,name,marks):
        self.name = name
        self.marks = marks
        
    def show(self):
        print(self.name,"\n", self.marks)
        
rev = revision("David", 95)
rev.show()
```

```python
class calculator:
    def __init__(self, num1, num2):
        self.num1 = num1
        self.num2 = num2
        
    def Add(self):    # Method - Add()
        print(self.num1 + self.num2)  
        # print(self.num1 - self.num2) 
    
    def Sub(self):
        print(self.num1 - self.num2) 
        
cal = calculator(10,50)
cal.Add()
print()
cal.Sub()
    
```
