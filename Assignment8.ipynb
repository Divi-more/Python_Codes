### Dictionary

1. Write a program to count the frequency of each character in a string using a dictionary. 
- Input: "programming" 
- Output: {'p':1, 'r':2, 'o':1, ...} 

```python
str1 = "Programming"
print("String is :", str1)

frequency = {}              # empty dictionary

for i in str1:              # iterate the string
    if i in frequency:      # if it exist in freqency already, 
        frequency[i] += 1   # increase the count of the character by 1.
    else:
        frequency[i] = 1    # otherwise, add it to srequency dict
print("Frequency of each character: \n",frequency) 
```

2. Write a program to count the frequency of each element in a list using a dictionary.
- Input: [10,20,10,30,20,10] 
- Output: {10:3, 20:2, 30:1}

```python
list1 = [10,20,10,30,20,10]
print("List: ",list1)

frequency = {}

# Count each character
for i in list1:
    if i in frequency:
        frequency[i] += 1
    else:
        frequency[i] = 1

# Display the result
print("Frequency of each element: \n",frequency)
```

3. Write a program to find the key having the maximum value in a dictionary.
- Input: {'A':85, 'B':92, 'C':78} 
- Output: B 

```python
dict1 = {'A':85, 'B':92, 'C':78}
print("Dict: ", dict1)

for i in dict1:
    if dict1[i] == max(dict1.values()):
        print("Key with max value:",i)
    
```

6. Write a program to remove duplicate values from a dictionary. 
- Input: {'a':10, 'b':20, 'c':10, 'd':30} 
- Output: {'a':10, 'b':20, 'd':30} 

```python
dict1 = {'a':10, 'b':20, 'c':10, 'd':30}
print(dict1)

duplicate = set()
result = {}

for key, value in dict1.items():
    if value not in duplicate:
        duplicate.add(value)
        result[key] = value

print("After removing duplicates:", result)
        
```

7. Write a program to check whether a given key exists in a dictionary. 

```python
dict1 = {'a':10, 'b':20, 'c':30, 'd':40}
print(dict1)

k=input("Enter key to check(a-d): ")
found = False

for i in dict1.keys():
    if i == k:
        found = True
        break
if found:
    print("Exist in dictionary")
else:
    print("Not exist in dictionary")

```

8. Write a program to find the sum of all values in a dictionary. 
- {'Math':85, 'Science':90, 'English':80} 
- Output: 255

```python
dict1 = {'Math':85, 'Science':90, 'English':80} 
print("Dictionary: ", dict1)

res = 0
for i in dict1.values():
    res = res + int(i)
print("Addition of all values in dictionary: ",res)
    
```

9. Write a program to create a dictionary from two lists: one containing keys and the other containing values. 
- Keys = ['A', 'B', 'C'] 
- Values = [10, 20, 30]

```python
keys = ['A', 'B', 'C'] 
values = [10, 20, 30]
print("Keys: ",keys)
print("Values: ", values)

dict1 = {}
for i in keys:
    for j in values:
        dict1[i] = j
print(dict1)
```

10. Write a program to create a student result dictionary containing name and marks of 5 students, then display the topper. 

```python
students = {}

for i in range(5):
    name = input("Enter student name: ")
    marks = int(input("Enter marks: "))
    students[name] = marks

topper = max(students, key=students.get)

print("\nStudent Result Dictionary:")
print(students)

print("\nTopper:")
print("Name:", topper)
print("Marks:", students[topper])
```
