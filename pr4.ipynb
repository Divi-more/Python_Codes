String Data Type : 
    - (" "), (' ')
* Properties:
    1. ordered
    2. immutable
    3. indexed
    4. slicing supported
    5. Homogeneous- stores only character not numbers. but if nums are as string then yes
    6. Sequence type
    7. Support +ve and -ve indexing
    8. Concatination support - join 2 strings
    9. Dynamic memory allocation
    10. Allow Diplicates
    11. Repetation
    12. Iterable


```python
str_demo = "Hello"
name1 = 'Divya'
print(str_demo, type(str_demo))
print(name1)

info = """Hello !
Divya
More 
          Here"""
          
print(info)

# Unordered data type 'set'
set1 =  {5, 4, 9, 2}
print(set1)

# immutable- it saved 'a' as two separate vars in 2 different locations
a = "A"
print(a, id(a))
a = "B"
print(a, id(a))

# indexed
name = "usernamedivya"
# print(name[3])

# Slicing support- can access part of string

print("\n",name[8: 13: 1]) #[start : end: step]
print(name[4 : 8 ]) 
print(name[0:])
print(name[:4])
print(name[:])
print(name[::-1]) #reverse

# Homogeneous
a = "abc123"
print("\n",a)

# +ve and -ve index support
# 01234      +ve index      left to right(|----------->)
# -4-3-2-1   -ve index      right to left(<-----------|)
print("\n",name[8])
print(name[-4])

# Concatination
print( str_demo +" "+ name1)
```

```python
#iterate string using for loop
# name = "usernamedivya"
for i in name:
    print(i, end=" ")
        
```

String Methods:
    1. upper()
    2. lower()
    3. capitalize()
    4. title()
    5. swapcase()- lower -> upper and vice versa
    6. strip() - removes white spaces from start or end of string
    7. replace(old, new)
    8. split()
    9. join()
    10. find()
    11. index() 
    12. count()
    13. startswith() - return T or False
    14. endswith()
    15. isalpha() - all characters
    16. isanum() - only nums, only characters, mixed characters and number
    17. isdigit() - all nums only

```python

str1 = "DEMO"
str2 = "user"

print(str2.upper()) 

print(str1.lower())

str3 = "demo user is using system..."
print(str3.capitalize()) 

print(str3.title())

print(str1.swapcase())
print(str2.swapcase())

temp = "   demo      "
print(temp)
print(temp.strip())
print(temp.rstrip())    # right side space removed
print(temp.lstrip())    # left ========||========

print("\n\n",str3.replace("demo", 'Divya'))

print(str3.split())
dm = "demo, user is, using system"
print(dm.split(","))

listone= ['demo', 'user', 'is', 'using', 'system...', '10']
temp = " ".join(listone)
print("\n\n",temp)
print(" ".join(listone),"\n\n")

print(temp.find("using"))
print(temp.find("o"))
print(temp.find("h")) # doesn't exist and also work for lower and upper charaters

print("\n\nindex()",temp.index("o"))

print("\n\n", temp.count("e")) # how many time 'e' come in string 

print(temp.startswith("d"))
print(temp.startswith("D"))
print(temp.endswith("e"))
print(temp.endswith("0"))

str4 = "demo123"
print("\n\n",str1.isalpha())
print(str4.isalpha())
print(str4.isalnum())
print(str1.isalnum())
print(str4.isdigit())

```

Built -in function: predefined functions:
    1. len() - returns length of string
    2. max() - return maximum value
                ( if it's numbers, then it return largest digit.
                  but for string it check their ASCII value and return max)
    3. min() - return minimum based on ASCII values
    4. sorted() - sort the string in ascending order
    5. type() - return datatype
    6. list() - 
    7. tuple() -
    8. set() - 
    9. str() - convert to string
    10. enumerate() - gives object value and index
    11. reverse()
    12. ord() - return ASCII value
    13. chr() - return char at ASCII value
    14. any() - 
    15. all() - 
    `14 15 work properly with list, tuple, set`


```python
str1 = "demo121"
str2 = "12734"

print(len(str1))

print(ord("d"), ord("e"), ord("m"), ord("o"), ord("1")) # give ASCII value of letters
print(chr(100)) # give letter of ASCII value
print(max(str1))
print(max(str2))

print(min(str1))

print(sorted(str2))
print(sorted(str1))

print(type(str1))
print(type(str2))

print(list(str1))

print(tuple(str2))

print(set(str1))

print(str(str1))

for i, j in enumerate(str1):
    print(i, j)
print(enumerate(str2))

print(reversed(str1))
print(" ".join(reversed(str1)))

list1 = ['demo', 'user', False]
print(any(list1))

print(all(list1))


```
