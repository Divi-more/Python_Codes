1. Write a program to count the number of vowels and consonants in a given string. 

```python

str1 = input("Enter string: ")
print("Input string: ", str1)
 
vowels = 0
consonents = 0

for i in str1:
    if i.isalpha():                         # check if all are characters or not, if yes then inner loop
        if i.lower() in 'aeiou':            # convert string into lowercase and checks for vowels. if vowels are found
            vowels += 1                     # increases vowel count by 1
        else:
            consonents += 1                 # if not, then increases consonent count by 1
            
print("Number of Vowels: ", vowels)
print("Number of Consonents: ", consonents) 
```

2. Write a program to check whether a given string is a Palindrome. 
- Input: madam 
- Output : Palindrome 

```python
str1 = input("Enter a string: ")
print("String to check: ", str1)

str1 = str1.lower()
str2 = str1[: : -1]
print("Reversed string: ",str2)

if str1 == str2:
    print(str1," is Palindrome.")
else:
    print(str1," is not Palindrome.")
```

3. Write a program to reverse a string without using slicing ([::-1]) or built-in functions. 

```python
str1 = input("Enter a string: ")
print("String is: ",str1)

reverse = ""

for i in str1:
    reverse = i + reverse
    
print("Reversed string: ", reverse)        
```

6. Write a program to find the longest word in a sentence. 
- Input: Python is an amazing programming language  
- Output: programming

```python
str1 = input("Enter a sentence: ")
print("Sentence: ", str1)

str1 = str1.strip().lower()

words = str1.split()

longest = ""
for i in words:
    if len(i) > len(longest):
        longest = i 
print("Longest word: ", longest)
    

# max_word = max(words, key=len)
# print("Longest word: ", max_word )
    
```

7. Write a program to capitalize the first letter of every word without using the title() method. 
- Input:    welcome to python 
- Output:   Welcome To Python 

```python
str1 = input("Enter a sentence: ")
print("Sentence: ", str1)

words = str1.strip().lower().split()
print(words)

for word in words:
    print(word.capitalize(), end=" ")
        
```

8. Write a program to count the total number of words in a sentence without using the split() method. 
- Input:     Python is easy to learn 
- Output:    5 

```python
str1 = input("Enter a sentence: ")
print("Sentence: ", str1)

str1 = str1.strip().lower()

count = 1

for i in str1:
    if i.startswith(" "):
        count += 1
print("Total count of words: ",count)
```
