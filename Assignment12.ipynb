1. What will be the output of the following code? Explain which namespace Python uses for x.  

```python
x = 10  

def test():     
    x = 20     
    print(x)  
    
test()
print(x) 

print("Expected Answer: \n20\n10") 
```

2. What will be the output? Explain the LEGB rule step by step.

```python
x = "Global"                # global namespace 

def outer():     
    x = "Enclosing"         # local namespace  
    
    def inner():         
        print(x)            # Enclosing - shadowing concept  
        
    inner()  
    
outer()

print("Expected Answer: \nEnclosing")
```

3. Identify the Local, Enclosing, Global, and Built-in names in the following program:

```python
x = 10  

def outer():     
    y = 20      
    
    def inner():         
        z = 30         
        print(x, y, z, len("Python"))      
        
    inner()  
    
outer()

print(f"\nGlobal: 10 \nLocal: 30 \nEnclosing: 20 \nBuilt-in: len() [6]")
```

4. What will be the output? Is x being shadowed?  

```python
"""
Yes, x is beinged shadowed.
o/p:- 300
      200
      100
"""

x = 100                 # Global

def outer():     
    x = 200             # Enclosing
    
    def inner():         
        x = 300         # Local     
        print("Local: ",x)      
        
    inner()     
    print("Enclosing: ",x)  
    
outer() 
print("Global: ",x) 
```

5. What will happen when this program runs? Explain why. 

```python
count = 10              # global namespace

def update():     
    count = count + 1   # here count is being treated as local namespace,  
    print(count)        # so it won't be accessed
    
update() 

print(f"\n\n It'll raise a error")
```

6. Correct the following program using the global keyword so that the final output is 20.  

```python
x = 10  

def update(): 
    global x        # add this one line so it'll give o/p 20  
    x = 20  

update() 
print(x) 
```

7. What will be the output? Explain the difference between the Local and Enclosing Namespace.  

```python
def outer():     
    x = 10      
    
    def inner():         
        x = 20         
        print("Local namespace: ",x)      
        
    inner()     
    print("Enclosing namespace: ",x)  
    
outer() 

print("Local: 20 \n Enclosing: 10")
print("Local namespace is in inner function. It's accessible only for inner() in this program.")
print("Enclosing namespace is accessible for outer() only. \nThere's no Global namespace or variable declared in this program ")
```

8. What will be the output? Why does nonlocal change the value of x?  

```python
def outer():     
    x = 10      
    
    def inner():         
        nonlocal x         
        x = 50      
        
    inner()     
    print(x)  
    
outer() 
print("Expected Output: 50\n\n nonlocal x tell inner() to use existing x=10 from outer().")
print("x = 50 overwrite 10 value.")
```
