# Dictionary: store data in { key:value}

Characteristics / Properties:
    1. mutable - values can repete, not keys
    2. ordered - maintain sequence of creation
    3. key-value pair
    4. key ust be unique - duplicate keys not allowed
    5. value can be duplicate
    6. indexed by key
    7. iterable
    8. heterogenous-(different data types can be stored) & homogeneous-(same datatype only)
    9. Dynamic - size can change
    10. nested dictionary support
    11. no numeric index
    12. no slicing
    13. fast searching, uses hashing techniques
    14. key must be immutable - keys can't be changed

# Methods:

    1. keys() - return all keys
    2. values() - return all values
    3. items() - get both values and keys
    4. get() - get value by key
    5. update() - to add multiple values value
    6. pop() - remove specified key
    7. popitem() - removes last item
    8. clear() - remove all key-value pairs
    9. copy() - 
    10. count()
    11. type()

```python
dict1 = {1 : "user1", 2 : "user2", 3 : "user3", 4:"user4"} 
print("Dict 1: ",dict1)

print(dict1.keys())         # can be used for iteration or for loop

for i in dict1.values():
    print(i)
    
for i in dict1.keys():
    print(i)
    
for i in dict1.items():
    print(i)
    
print(dict1.get(2))

dict1[5]="user5"    #add 1 value only
dict1.update({6:"user6",7:"user7"})
print(dict1)

print(dict1.pop(7))
print("After popping : ",dict1)

dict1.popitem()
print("Pop items() : ",dict1)

# dict1.clear()
# print("Pop items() : ",dict1)

dictCopy = dict1.copy()
print("Copied: ",dictCopy)

```

```python
dict1 = {1 : "user1", 2 : "user2", 3 : "user3", 4:"user4", 0:0} 
# print("Dict 1: ",dict1[1])

print("Length: ",len(dict1))

print("Max:", max(dict1))

print("Min:", min(dict1))

print("sorted: ", sorted(dict1))

# print(type(dict1))

# print(list(dict1))

# tup = tuple(dict1)
# print(tup)

# print(set(dict1))

# str1 = str(dict1)
# print(str1[0])

print("index: ", "Keys: ", end=" ")
for i,j in enumerate(dict1):
    print("\n",i, "     ",j)
    

print("\n","index: ", "Values: ", end=" ")
for i,v in enumerate(dict1.values()):
    print("\n",i, "     ",v)
    

print(any(dict1))

print(all(dict1))

```
