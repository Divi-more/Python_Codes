### Exception Handling

1. Write a program to handle ZeroDivisionError when dividing two numbers entered by the user. 

```python
num1 = int(input("Enter num 1: "))
num2 = int(input("Enter num 2: "))

print(f"Num1: {num1} \nNum2: {num2} \n")

try:
    res = num1/ num2
    
except ZeroDivisionError as e:
    print("Error: ", e)
    
else:
    print("Result: ", res)    
```

2. Write a program to handle ValueError when the user enters a non-numeric value for age.  

```python
try:
    age = int(input("Enter your age: "))
    
except ValueError as e:
    print(f"Error: {e}")
    
else:
    print(f"Age: {age}")
```

3. Write a program to handle IndexError while accessing an element from a list.  

```python
list1 = [10,20,30,40,50]
print(f"List: {list1} \n")

try:
    print(list1[6])
    
except IndexError as e:
    print(f"Error: {e}")
    
else:
    print(f"Index: {list1[6]}")
```

4. Write a program to handle KeyError when accessing a key from a dictionary.  

```python
dict1 = {1 : "A", 2 : "B", 3 : "C", 4 : "D"}
print(f"Dictionary: {dict1}\n")

try:
    key = int(input("Enter a key to access: "))
    print(f"Value: {dict1[key]}")
    
except KeyError as e:
    print("KeyError: The entered key does not exist in the dictionary.")
```

5. Write a program to handle TypeError by attempting to add an integer and a string.  

```python
str1 = input("Enter a string: ")
print(f"String: {str1} ")
num1 = int(input("Enter a num : "))
print(f"Num: {num1}\n")

try:
    res = str1 + num1

except TypeError as e:
    print(f"Error: {e}")
    
else:
    print(f"Result: {res}")
```

6. Write a program that demonstrates the use of try, except, else, and finally blocks.  

```python
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
print(f"Num1: {num1} \nNum2: {num2}")

try:
    result = num1 / num2

except ZeroDivisionError as e:
    print(f"Error: {e}")

else:
    print(f"Result: {result}")

finally:
    print("Program execution completed.")
```
