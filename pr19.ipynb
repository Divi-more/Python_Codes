# Aggregation:-
    - we use object to access properties of prev class into new class without inheritance.
    - pass 1 obj as arguement for another obj

```python
class result:
    
    def __init__(self, per, res):
        self.percentage = per
        self.res = res
        
    # def display(self):
    #     print(f"Percentage: {self.percentage}\nResult: {self.res} ")

class student:
    
    def __init__(self, name, result):
        self.name = name
        self.result = result
        
    def display(self):
        print(f"Name: {self.name}\nResult: {self.result.percentage}\n{self.result.res}")
        
obj = result(70, "Pass")  


obj1 = student("Divya", obj) 
# print(obj1.result)       
obj1.display()
```

```python
class address:
    
    def __init__(self, city, pin):
        self.city = city
        self.pin = pin

class student:
    def __init__(self, name, add):
        self.name = name
        self.add = add
        
    def display(self):
        print(self.name)
        print(self.add.city)
        print(self.add.pin)
        
        
obj1 = address("Pune", 422103)         
        
obj2 = student("Divya", obj1)
obj2.display()
```

# Method Overriding / Polymorphism:-
 - same name method in diff classes, obj pgive preference to its own class method.
 - i such cases we've to create obj for each class and then access each method of it's own class.
 - it's also known as "Polymorphism"

```python
class upi:
    def pay(self):
        print("UPI Payment")
    
class netbankikng(upi):
    def pay(self):
        print("Net Banking Payment")
    
class creditcard(netbankikng):
    def pay(self):
        print("Credit Card Payment")
        
obj1 = creditcard()
obj1.pay()

obj2 = netbankikng()
obj2.pay()

obj3 = upi()
obj3.pay()
```

# Encapsulation:-
    - getter - setter
    - hide private data

```python
class bank:
    def __init__(self):
        self.name = "Demo"
        self.__pin = 123456
        
    def get_pin(self):          # getter: we can't access private var in other class, but can access method, so we create  getter method
        return self.__pin
        
class subbranch(bank):
    def display(self):
        print(self.name)
        print(self.get_pin())
        
obj1 = subbranch()
obj1.display()
```

```python
class bank:
    def __init__(self):
        self.name = "Demo"
        self.__pin = 123456
        
    def get_pin(self):          
        return self.__pin
    
    def set_pin(self, newpin):  # setter: to update private variable's value
        strpin = str(newpin)
        
        if len(strpin) >= 6: 
            self.__pin = newpin
        else:
            print("Enter 6 digit password ")
        
class subbranch(bank):
    def display(self):
        print(self.name)
        print("Old pass: ",self.get_pin())
        
        self.set_pin(152012)      # call setter
        print("New pass: ",self.get_pin())
        
obj1 = subbranch()
obj1.display()
```
