1. Write a Python program using nested loops to print the following pattern:  
    1 
    12 
    123 
    1234 
    12345 

```python
for i in range(1, 6):
    for j in range(1, i + 1):
        print(j, end="")
    print()
```

2. Write a Python program using nested loops to find and print all prime numbers between 1 and 100.  

```python
for num in range(2, 101):
    prime = True

    for i in range(2, num):
        if num % i == 0:
            prime = False
            break

    if prime:
        print(num, end=" ")
```

3. Given two lists, use nested loops to find and print the common elements:  list1 = [10, 20, 30, 40, 50] list2 = [20, 40, 60, 80, 100] 

```python
list1 = [10, 20, 30, 40, 50] 
list2 = [20, 40, 60, 80, 30, 100] 

for i in list1:
    if i in list2:
        print(i, end=" ")
```

4. Write a Python program to count the frequency of each character in a string without using count().  

```python
str1 = "Hello World"

frequency = {}   

for i in str1:
    frequency[i] = frequency.get(i, 0) + 1

print(frequency)
```

5. Write a Python program to find the first non-repeated character in a string.  

```python
str1 = "Programming"

for i in str1:
    count = 0
    
    for j in str1:
        if i == j:
            count += 1

    if count == 1:
        print(i)
        break
```
