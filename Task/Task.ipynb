2. Write a Python program using nested loops to find and print all prime numbers between 1 and 100.  

```python
for num in range(2, 101):
    prime = True

    for i in range(2, num):
        if num % i == 0:
            prime = False
            break

    if prime:
        print(num, end=" ")
```

10. Given two dictionaries containing product names and prices, merge them into one dictionary. If a product exists in both, keep the higher price.  

```python
dict1 = {"Laptop": 80000, "Mobile": 20000, "Smartwatch": 15000, "Tablet": 30000, "Desktop": 30000}
dict2 = {"Mobile": 20000, "Tablet": 35000, "Desktop": 40000}
print(f"Original Dictionaries: \n\n{dict1} \n{dict2}")

for i, j in dict2.items():
    if i in dict1:
       if j > dict1[i]:
           dict1[i] = j
       else:
           dict1[i] = j

print(f"\nUpdated Dictionary: \n{dict1}")
```

11. Create a function calculator(a, b, operation) that performs addition, subtraction, multiplication, and division based on the given operation. 

```python
def calculator(num1, num2, operator):
    if operator == "+":
        return num1 + num2
    elif operator == "-":
        return num1 - num2
    elif operator == "*":
        return num1 * num2
    elif operator == "/":
        return num1 / num2
    else:
        return "Invalid Operator"
    
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
operator = input("Enter operator: ")

print(f"Num1: {a}, Num2: {b}, Operator: {operator}")
print("Result: ",calculator(a, b, operator))
```

12. Create a function that accepts a list of numbers and returns the largest number, smallest number, sum, and average.

13. Create a BankAccount class with account holder, account number, and balance. Add methods for deposit, withdrawal, and displaying balance. Prevent withdrawal when the balance is insufficient.  

```python
class BankAccount:
    def __init__(self, account_number, account_holder, balance):
        self.account_number = account_number
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited: {amount}. New Balance: {self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds")
        else:
            self.balance -= amount
            print(f"Withdrew: {amount}. New Balance: {self.balance}")

    def display_balance(self):
        print(f"Account Number: {self.account_number}, Account Holder: {self.account_holder}, Balance: {self.balance}")
        
obj = BankAccount(12345, "Divya", 50000)

obj.display_balance()

obj.deposit(5000)

obj.withdraw(10000)

obj.display_balance()
```

14. Create an Employee class with employee name, ID, department, and salary. Add methods to display details, increase salary by a given percentage, and check whether the salary is above ₹50,000.  

```python
from hmac import new


class Employee:
    def __init__(self, name, emp_id, dept, salary):
        self.name = name
        self.emp_id = emp_id
        self.dept = dept
        self.salary = salary

    def display_details(self):
        print(f"Employee Name: {self.name} \nEmployee ID: {self.emp_id} \nDepartment: {self.dept} \nSalary: {self.salary}")

    def update_salary(self, new_salary):
        self.salary = self.salary + (self.salary * new_salary / 100)
        print(f"Salary after {new_salary}% increase: ₹{self.salary}")
        
        
    def check_salary(self):
        if self.salary > 50000:
            print(f"{self.name} has a good salary of ₹{self.salary}")
        else:
            print(f"{self.name} has a low salary of ₹{self.salary}")
    
obj = Employee("Divya", 101, "IT", 50000)

obj.display_details()
print()

obj.check_salary()
print()

obj.update_salary(20)
print()

obj.display_details()
print()

obj.check_salary()
```

15. Create a Vehicle class with brand and speed. Create Car and Bike child classes with additional attributes. Use inheritance and method overriding to display their complete details.  

```python
class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed

    def display_details(self):
        print(f"Vehicle Brand: {self.brand} \nSpeed: {self.speed}")
        
class Car(Vehicle):
    def __init__(self, brand, speed, model, year, color):
        super().__init__(brand, speed)
        self.model = model
        self.year = year
        self.color = color
        
    def display_details(self):
        super().display_details()
        print(f"Car Model: {self.model} \nYear: {self.year} \nColor: {self.color}")
        
class Bike(Vehicle):
    def __init__(self, brand, speed, type_bike, year, color):
        super().__init__(brand, speed)
        self.type_bike = type_bike
        self.year = year
        self.color = color
        
    def display_details(self):
        super().display_details()
        print(f"Bike Type: {self.type_bike} \nYear: {self.year} \nColor: {self.color}")
        
car = Car("Toyota", 180, "Camry", 2020, "Black")
bike = Bike("Yamaha", 150, "Sports", 2021, "Blue")

car.display_details()

print()

bike.display_details()
```
