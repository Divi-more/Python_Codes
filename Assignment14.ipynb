### 1. Student Information 
- System Create a Student class with: • name  • roll_no  • branch  • marks  
- Create 3 student objects and create a method display() to print their details. 

```python
class Student:
    def get_info(self, name, rollno, branch, marks):
        self.name = name
        self.rollno = rollno
        self.branch = branch
        self.marks = marks
        
    def display(self):
        print("Name: ", self.name)
        print(f"Roll No.: {self.rollno}")
        print(f"Branch: {self.branch}")
        print(f"Marks: {self.marks}")
        
std1 = Student()
std1.get_info("Divya", 101, "CS", 80)
std1.display()
print()
std2 = Student()
std2.get_info("Shubhangi", 121, "IT", 75)
std2.display()
print()
std3 = Student()
std3.get_info("Dipali", 103, "CS",70)
std3.display()
```

### 2. Employee Salary 
- Create an Employee class with: • name  • employee_id  • salary  
- Create a method display_salary() that displays the employee's salary. 
- Create 3 employee objects. : Don't assign the values directly inside the class. Pass them while creating objects. 

```python
class Employee:
    def __init__(self, name, emp_id, salary):
        self.name = name
        self.emp_id = emp_id
        self.salary = salary
        
    def display(self):
        print(self.name, self.emp_id, self.salary)
        
emp1 = Employee("Jay", 101, 10000)
emp2 = Employee("Priya", 102, 12000)
emp3 = Employee("Vishu", 103, 10000)
emp1.display()
emp2.display()
emp3.display() 
```

### 3. Mobile Phone 
- Create a Mobile class with: • brand  • model  • price  
- Create a method: display_details() 
- Create 3 mobile objects with different information. 
- Expected concept: Constructor + objects. 

```python
class Mobile:
    def __init__(self, brand, model, price):
        self.brand = brand
        self.model = model
        self.price = price
        
    def display(self):
        print(self.brand,self.price, self.model)
    
a = Mobile("Samsung", 25000, "Y58")
b = Mobile("Vivo", 20000, "Y27")
c = Mobile("Oppo", 24000, "O55")
a.display()
b.display()
c.display()
```
