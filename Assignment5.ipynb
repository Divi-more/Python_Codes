1. Write a program to count the number of even and odd numbers in a list. 

```python
listOne = [1,5,2,4,8,6,75,15,46]
print(listOne)

even = 0
odd = 0

for i in listOne:
    if i % 2 == 0:
        even += 1
    else:
        odd += 1    
print("Even numbers count: ", even)
print("Odd numbers count: ", odd)
```

2. Write a program to create a new list containing only the positive numbers from a given list. 
- Input: [-10, 20, -5, 40, 0, 15] 
- Output: [20, 40, 15] 

```python
listOne = [-10, 20, -5, 40, 0, 15]
print("Original List: ",listOne)

res = []
for i in listOne:
    if i > 0:
        res.append(i)
print("List containing only positive values from list: ",res)    
```

3. Write a program to remove all duplicate elements from a list while preserving the original order. 
- Input: [10,20,30,20,40,10,50]
- Output: [10,20,30,40,50] 

```python
listOne = [10,20,30,20,40,10,50]
print("Original List: ",listOne)

list1 = []

for i in listOne:
    if i not in list1:
        list1.append(i)
print("Updated by removing duplicate values: ",list1)        
        
```

4. Write a program to find the second largest element in a list.

6. Write a program to count the frequency of every element in a list. - Input: 
- [1,2,3,2,1,4,1] 
- Output: 1 -> 3 
          2 -> 2 
          3 -> 1 

```python
listOne = [1,2,3,2,1,4,1]
print("Original List: ",listOne)

for i in set(listOne):
    print(i, ":",listOne.count(i))
```

7. Write a program to reverse a list without using the reverse() method or slicing ([::-1]).

```python
listOne = [10,20,30,40,50]
print("Original List: ",listOne)

reverse = []

for i in listOne:
    reverse = [i] + reverse
    
print("Reversed list: ", reverse)        
```

8. Write a program to check whether a list is a Palindrome. 
- Input: [1,2,3,2,1] 
- Output: Palindrome 

```python
listOne = [1,2,3,2,1]
print("Original List: ",listOne)

if listOne == listOne[::-1]:
    print("Palindrome")
else:
    print("Not palindrome")    
    
```

9. Write a program to find all common elements between two lists. 
- List1 = [10,20,30,40] 
- List2 = [20,40,60] 
- Output: [20,40] 

```python
list1 = [10,20,3,40]
list2 = [20,40,60]

newlist = []
for i in list1:
    for j in list2:
        if i == j:
            newlist.append(i)
print(newlist)    
```

10. Write a program to separate all even and odd numbers into two different lists. 
- [10,15,20,25,30] 
- Output: Even = [10,20,30]
          Odd = [15,25] 

```python
list1 = [10,15,20,25,30]

even = []
odd = []

for i in list1:
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)
print("Even: ", even)
print("Odd: ", odd)
```
