### Function:

### Function: reusable block of code which can called anywhere in the program

    - Syntax:
        def function_name(parameters):  //Function initilazation
            #Body of function
        function_name(values)           //Function call

```python
#  Without Function

num1 = 10
num2 = 20

add = num1 + num2
print("The sum of", num1, "and", num2, "is:", add)

num1 = 15
num2 = 20 

add = num1 + num2
print("The sum of", num1, "and", num2, "is:", add)

num1 = 10
num2 = 50

add = num1 + num2
print("The sum of", num1, "and", num2, "is:", add)
```

```python
#  With Function

def add(num1, num2):
    res = num1 + num2
    print("The sum of", num1, "and", num2, "is:", res)

add(10, 20)
add(15, 20)
add(10, 50)
```

### Types of Function:

        1. Built-in function
        2. User-defined function

            i) No arguements, No return value
            ii) Arguements, No return value
            iii) No arguements, with return value 

```python
# No arguements, No return value

def temp():
    print("Hello World")
    
temp()

# Arguements, No return value
def greet(name):                # name is a parameter
    print("Hello", name)
    
greet("Divya")                  # Divya is an argument(value of parameter)

# No arguements, With return value
```
