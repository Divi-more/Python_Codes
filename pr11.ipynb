### Python Namespace

Namespace:- system used by python to store and manage names and their corresponding objects.
          - place where python keeps track of names like var, classes, functions
            (Can access it in only it's own scope)

Types:
built in
global
local
enclosing namespace  

1. Global Namespace: - 

```python
name = "Mohit"      # Global namespace (Can accessed anywhere in the program)
age = 22            # Global namespace


def temp():
    rollno = 55     # Local namespace
    mark = 75       # Local namespace
    print(f"Global:\n{name}\n{age}\n")
    print(f"Local:\n{rollno}\n{mark}\n")
        
temp()

print(f"Global:\n{name}\n{age}\n")
# print(f"{rollno}\n{mark}")  # rollno and mark are local, cann't be accessed globally
```

2. Local namespace:-

```python
def temp():
    username = "Mohit"      # Local namespace
    age = 22
    print(f"{username}\n{age}")

# print(f"{username}\n{age}")    
temp()

```

3. Enclosing namespace:-
    - namespace of an outer function that contains a nested(inner) function

```python
names = "Divya"             # Global namespace

def outer():
    age = 20                # Enclosing namespace (at 1st it'll be local, but once we create another function inside this 
                            #     then it becomes enclosing rather than local as another becomes local)
    
    # print(marks) #canno be accessed outside inner(), it doesn't fit the scope
    
    def inner():
        marks = 85          # Local namespace
        
        print(marks)
        print(age)
        print(names)
        
    inner()
    
outer()
```

4. Built-in namespace:-
    - predefined by python (keywords, functions)

### Shadowing:-
    - same variable declared in 2 diff scopes, both will be printed
    - same var "name", mohit is global and demo is local. in o/p, we got both 
    - creating local var of global var is Shadowing 

```python
name = "Mohit"

def temp():
    name = "demo"       # shadowing
    print(name)

temp()

print(name)
```

### Global as Keyword:-
    - using global as keyword avoid creating shadow of var.
    - it update globally defined value with new as shown in example below:

```python
name = "Mohit"

def temp():
    global name     # avoid shadowing, and update name wth "demo"
    name = "demo"
    print(name)

temp()

print(name)
```

### LEGB:- prefrence
    - Local: inside current function
    - Enclosing - out in outer function
    - Global - 
    - Built-in

```python
name = "demo"           # Global

def one():
    name = "demo1"      # Enclosing
    
    def two():
        name = "demo2"  # Local
        print("Local: ",name)   # demo2- local
    
    two()
    print("Enclosing: ",name, len(name))       # demo1 - enclosing
    
one()
print("Global: ",name)           # demo - global
print("Built-in: ",len(name))
```
