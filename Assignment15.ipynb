1. Student Class – 
__str__() Create a Student class with name, roll_no, and marks. Implement __str__() so that printing them. 

```python
class Student:
    def __init__(self, name, roll, marks):
        self.name = name
        self.roll = roll
        self.marks = marks
        
    def __str__(self):
        return f" Name: {self.name}\n Roll No: {self.roll}\n Marks: {self.marks}"
    
obj1 = Student("Divya", 101, 85)
print(obj1)
```

2. Product Class – __eq__() Create a Product class with name and price. Implement __eq__() to check whether two products have the same price. 

```python
class Product:
    
    def __init__(self,name,price):
        self.name = name 
        self.price = price
        
    def __eq__(self, value):
        return self.price == value.price
    
obj1 = Product("Laptop", 50000)
obj2 = Product("TV", 50000)

print(obj1 == obj2)
```

3. Shopping Cart – __len__() Create a ShoppingCart class containing a list of products. Implement __len__() so that len(cart) returns the number of products in the cart.

```python
class ShoppingCart:
    
    def __init__(self):
        self.product = []
        
    def add_products(self,product):
        self.product.append(self.product)
        
    def __len__(self):
        return len(self.product)
    
obj = ShoppingCart()

obj.add_products("TV")
obj.add_products("Mobile")
obj.add_products("Laptop")
obj.add_products("EarBuds")

print(len(obj))
```

4. Bank Account – __str__() Create a BankAccount class with account holder name, account number, and balance. Implement __str__() to display account information when the object is printed. 

```python
class BankAccount:
    
    def __init__(self, name, number, balance):
        self.name = name
        self.number = number
        self.balance = balance
        
    def __str__(self):
        return f" Name: {self.name} \n Number: {self.number} \n Balance: {self.balance}"
    
obj = BankAccount("Divya", 843852452, 50000)
print(obj)
```

5. Number Class – __add__() Create a Number class that stores one number. Implement __add__() so that two objects can be added 

```python
class Number:
    
    def __init__(self, num):
        self.num = num
        
    def __add__(self, other):
        return self.num + other.num
    
obj1 = Number(10)
obj2 = Number(10)

print(obj1 + obj2)
        
```

6. Create a Number class and implement: __add__()  __sub__()  __mul__()  __truediv__()  __mod__() 

```python
class Number:
    def __init__(self, num):
        self.num = num

    def __add__(self, other):
        return self.num + other.num

    def __sub__(self, other):
        return self.num - other.num

    def __mul__(self, other):
        return self.num * other.num

    def __truediv__(self, other):
        return self.num / other.num

    def __mod__(self, other):
        return self.num % other.num


num1 = Number(20)
num2 = Number(6)

print(num1 + num2)
print(num1 - num2)
print(num1 * num2)
print(num1 / num2)
print(num1 % num2)
```
