# 1. Class:

# 2. Object:
    - instance of a class

# 3. Method:
    - user defined functions or method. 

```python
# Class

class car:
    
    def greet():    # method
        print("Hello")

obj = car       # object creation
obj.greet()     # call function using object
```

# 4. Constructor:
    - __init__(self) is used to define constructor.
    - don't have to call separately.
    - get called automatically when we create object

# 5. Attributes:
    - variables passed in constructor
    - name, color, price, millage are all attributes

```python
class car:
    def __init__(self, name, color, price, milage):
    # def __init__(self):
        self.name = name
        self.color = color
        self.price = price
        self.milage = milage
        # self.name = "Demo"
        # self.color = "red"
        # self.price = "15L"
        # self.milage = 20
        
    def display(self):
        print(self.name)
        print(self.color)
        print(self.price)
        print(self.milage)
        
obj = car("nexon","white","10L",25)
obj.display()

# obj1 = car()
# obj1.display()


#user input
print
cname = input("Enter car name: ")
ccolor = input("Enter color: ")
cprice = input("Enter price: ")
cmilage = int(input("Enter milage: "))

obj3 = car(cname, ccolor, cprice, cmilage)
obj3.display()
```

```python
#practice 

list1 = [1,2,3]
list2 = [4,5,6]

    # 1,2,3,4,5,6
    # if we combine these two lists, then find center
    
list3 = list1 + list2
print(list3)

length = len(list3)
middle = length // 2

if length % 2 == 0:
    print(list3[middle])
else:
    res = (list3[middle-1] + list3[middle]) / 2
    print(res)
          
```

```python
list1 = list(map(int, input("Enter element of list: ").split()))
print(f"Given List is: {list1}")

num = len(list1)    # 6
print("Length: ", num)

if num % 2 == 1:      # if len = odd
    mid = num // 2    # return index 
    print(list1[mid])   # 30
    
else:               # if len = even
    mid = list1[num // 2 - 1], list1[num // 2]
    print(mid)
    
    res = (list1[num // 2 -1] + list1[num // 2]) // 2
    print(res)
```
