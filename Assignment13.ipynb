1. Write a program to create a text file and store the name, age, and marks of a student in it. 

```python
name = input("Enter a name: ")
age = int(input("Enter your age: "))
marks = int(input("Enter your marks: "))

file = open("Demo.txt", "w")
file.write("Name: " + name + "\n")
file.write("Age: " + str(age) + "\n")
file.write("Marks: " + str(marks) + "\n")

file.close()
print("Student details saved successfully.")
```

6. Write a program to read a file and create another file containing only the lines that contain the word "Python". 

```python
with open("demo3.txt", "r") as file:
    data = file.readlines()
    
with open("Newfile.txt", "w") as newfile:
    for lines in data:
        if "python" in lines.lower():
            newfile.write(lines)
```

8. Write a program to read a file and find the longest word present in it.  

```python
with open("demo3.txt", "r") as file:
    data = file.read()

word = data.split()
longest = word[0]

for i in word:
    if len(i) > len(longest):
        longest = i

print("Longest word is: ", longest)
```

9. Write a program to read a file containing numbers and calculate their sum, average, maximum, and minimum.  

```python
with open("nums.txt", "r") as file:
    data = file.read()
    print("Data: ", data)
    
    nums = list(map(int, data.split()))    
    
adds = sum(nums)
avg = adds/len(nums)
maxm = max(nums)
minm = min(nums)
print(f"Addition: {adds}\nAverage: {avg}\nMaximum: {maxm}\nMinimum: {minm}")
```
