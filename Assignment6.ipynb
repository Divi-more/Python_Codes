1. Write a program to find the largest and smallest element in a tuple without using max() and min().

2. Write a program to count the number of even and odd numbers in a tuple. 

```python
tuple1 = (2,4,1,8,5,2,9,5,7,6,3)
print("Tuple is: ", tuple1)

even = 0
odd = 0

for i in tuple1:
    if i % 2 == 0:
        even += 1
    else:
        odd += 1
print("Total even numbers in tuple: ", even)
print("Total odd numbers in tuple is: ", odd)    
```

3. Write a program to find all common elements between two tuples.

```python
tuple1 = (10,20,30,40,50)
tuple2 = (15,50,20,40,10)
print("Tuple 1 is: ", tuple1)
print("Tuple 2 is: ", tuple2)

for i in set(tuple1):
    for j in set(tuple2):
        if i == j:
            print(i, end=" ")
```

4. Write a program to create a new tuple containing only positive numbers from a given tuple. 

```python
tuple1 = (10,-20,30,40,-50, -3, 54)
print("Tuple is: ", tuple1)

tuple2 = ()
for i in tuple1:
    if i >= 0:
        tuple2 += (i, )
print("Positive tuple: ",tuple2)
```

5. Write a program to remove duplicate elements from a tuple while preserving the original order. 

```python
tuple1 = (1,3,2,5,4,2,4,3,6,8,9,5,7,8)
print("Tuple is: ", tuple1)

new = ()

for i in tuple1:
    if i not in new:
        new += (i,)
print("Tuple after removing duplicate tuples: ",new)   
```

6. Write a program to separate all even and odd elements of a tuple into two different tuples. 
- Input: (10,15,20,25,30) 
- Output: Even = (10,20,30) 
          Odd = (15,25) 

```python
tuple1 = (1,3,2,5,4,2,4,3,6,8,9,5,7,8)
print("Tuple is: ", tuple1)

even = ()
odd = ()

for i in tuple1:
    if i % 2 == 0:
        even += (i,)
    else:
        odd += (i,)
print("Even nums in tuple: ",even)
print("Odd nums in tuple: ",odd)
```

7. Write a program to find the intersection of two tuples. 

```python
tuple1 = (10,20,30,40,50)
tuple2 = (15,50,20,40,10)
print("Tuple 1 is: ", tuple1)
print("Tuple 2 is: ", tuple2)

print("Intersection: ", end=" ")
for i in set(tuple1):
    for j in set(tuple2):
        if i == j:
            print(i, end=" ")
            
```

8. Write a program to find the union of two tuples (unique elements only).

```python
tuple1 = (10,20,30,40,50)
tuple2 = (15,50,20,40,10)
print("Tuple 1 is: ", tuple1)
print("Tuple 2 is: ", tuple2)

union = tuple1

for i in tuple2:
    if i not in union:
        union += (i, )
print("Union: ",union)
```
